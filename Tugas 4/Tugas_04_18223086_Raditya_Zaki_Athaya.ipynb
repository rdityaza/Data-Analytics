**1. Analisis Korelasi Sederhana**

```python
import pandas as pd

# Load the CSV file into a pandas DataFrame
df = pd.read_csv('/content/coffee_sales_correlation_dataset.csv')

# Calculate the mean of each column
column_means = df.mean(numeric_only=True)

# Display the column means
display(column_means)
```

```python
# Pandas punya fungsi .corr() yang otomatis menghitung Pearson Correlation
# Menggunakan nama kolom yang sesuai dengan dataset (df)
korelasi_matrix = df[['Temperature_C', 'Rainfall_mm', 'Advertising_Budget', 'Coffee_Sales']].corr()

# Mengambil kolom 'Coffee_Sales' saja untuk ditampilkan
display(korelasi_matrix[['Coffee_Sales']])
```

**2. Visualisasi Hubungan Variabel**

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Mengatur gaya visual
sns.set_theme(style="whitegrid")

# Grafik 1: Suhu Udara vs Penjualan Kopi
plt.figure(figsize=(7, 4))
sns.scatterplot(data=df, x='Temperature_C', y='Coffee_Sales', color='darkorange', s=80)
plt.title('Suhu Udara vs Penjualan Kopi', fontsize=12, fontweight='bold')
plt.xlabel('Suhu Udara (°C)', fontsize=10)
plt.ylabel('Jumlah Kopi Terjual (cup)', fontsize=10)
plt.show()

print("\n")

# Grafik 2: Curah Hujan vs Penjualan Kopi
plt.figure(figsize=(7, 4))
sns.scatterplot(data=df, x='Rainfall_mm', y='Coffee_Sales', color='steelblue', s=80)
plt.title('Curah Hujan vs Penjualan Kopi', fontsize=12, fontweight='bold')
plt.xlabel('Curah Hujan (mm)', fontsize=10)
plt.ylabel('Jumlah Kopi Terjual (cup)', fontsize=10)
plt.show()
```
